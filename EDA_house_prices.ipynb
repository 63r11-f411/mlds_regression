```python
missing_cols = df.columns[df.isna().mean() > 0]

missing_info = pd.DataFrame({
    "dtype": df[missing_cols].dtypes,
    "missing_pct": df[missing_cols].isna().mean() * 100
})

missing_info["kind"] = missing_info["dtype"].apply(
    lambda dt: "numerical" if pd.api.types.is_numeric_dtype(dt) else "categorical"
)

missing_info = missing_info.sort_values(["kind", "missing_pct"], ascending=[True, False])

print("Categorical missing fields:")
print(missing_info[missing_info["kind"] == "categorical"].index.tolist())
print()
print("Numerical missing fields:")
print(missing_info[missing_info["kind"] == "numerical"].index.tolist())
```

## P.S. Решения по колонкам для FE

Нашлись позже, во время работы над кодом. Цифры в ячейке ниже.

- Удаляем `PoolQC` (всего 7 бассейнов), `Utilities` (почти константа), `GarageYrBlt` (у 75% домов совпадает с `YearBuilt`).
- NaN в порядковых признаках, площадях и счётчиках означает «объекта нет», поэтому → 0.
- `Functional` NaN → `'Typ'`, так сказано в описании данных.
- `MSSubClass` → строка: по описанию данных это код типа жилья, а не число.
- У пары домов объект есть, но одна характеристика не записана. Они тоже получают 0, это сознательное упрощение.

```python
print('Pools:', (df['PoolArea'] > 0).sum())
print('Utilities:', df['Utilities'].value_counts().to_dict())
print('GarageYrBlt == YearBuilt:', (df['GarageYrBlt'] == df['YearBuilt']).mean().round(2))
print('GarageYrBlt vs YearBuilt correlation:', round(df['GarageYrBlt'].corr(df['YearBuilt']), 2))

# NaN in a feature almost always means the object itself is absent
for feature, size in [('BsmtQual', 'TotalBsmtSF'), ('BsmtExposure', 'TotalBsmtSF'), ('BsmtFinType2', 'TotalBsmtSF'),
                      ('GarageQual', 'GarageArea'), ('FireplaceQu', 'Fireplaces')]:
    missing = df[feature].isna()
    print(f'{feature}: {missing.sum()} NaN, object present in {(missing & (df[size] > 0)).sum()}')
print('MasVnrArea NaN:', df['MasVnrArea'].isna().sum(), ', MasVnrType also NaN:', (df['MasVnrArea'].isna() & df['MasVnrType'].isna()).sum())
```
